```python

x = df.drop(['show_id', 'rating'], axis=1)
y = df['rating']
x['date_added'] = pd.to_datetime(x['date_added'], format='mixed', errors='coerce')
x['month_added'] = x['date_added'].dt.month
x['year_added'] = x['date_added'].dt.year
x['month_added'] = x['month_added'].fillna(0)
x['year_added'] = x['year_added'].fillna(0)
x = x.drop('date_added', axis=1)
def get_duration_numeric(duration_str):
    if pd.isna(duration_str):
        return 0
    if 'min' in duration_str:
        return int(duration_str.replace(' min', ''))
    elif 'Season' in duration_str:
        return int(duration_str.split(' ')[0])
    return 0
x['duration'] = x['duration'].apply(get_duration_numeric)
for col in x.select_dtypes(include='object').columns:
    x[col] = LabelEncoder().fit_transform(x[col].astype(str))
x = x.fillna(0)
y = LabelEncoder().fit_transform(y.astype(str))
display(x.head())
x.info()
```

```python

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print("x_train shape:", x_train.shape)
print("x_test shape:", x_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print("Data split into training and testing sets.")
print("x_train shape:", x_train.shape)
print("x_test shape:", x_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(x_train, y_train)

y_pred = model.predict(x_test)

print("Model trained successfully!")
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```
